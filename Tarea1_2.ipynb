```python
# c) Función 

def calculo_energia_cinetica (m_kg, v_kmh):
    v_ms= v_kmh / 3.6

    E_k= 0.5 *m_kg * (v_ms**2)
    return E_k 

masa= 4
velocidad= 2

resultado= calculo_energia_cinetica (masa, velocidad) 
print("La energía cinética del objeto es:",resultado, "Joules")
```
